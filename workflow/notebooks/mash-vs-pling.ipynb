```python
import string

import pandas as pd
import yaml
```

```python
df = pd.read_csv(snakemake.input[0])
```

```python
df = df[df["accession"].str.startswith("pRIVM") & (df["group_pling"] == "Singleton") & (df["group_mash"] != "Singleton")]
```

```python
subplots = {str(a) : b for a, b in  zip(sorted(df["group_mash"].astype(int).unique()), string.ascii_uppercase)}
```

```python
df["subplot"] = df["group_mash"].map(subplots)
```

```python
samples = df.melt(id_vars="subplot", value_vars=["accession", "nearest_neighbour"]).groupby("subplot")["value"].apply(list).to_dict()
```

```python
with open(snakemake.output[0], "w") as file:
    yaml.safe_dump(samples, file)
```
