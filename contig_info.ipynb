```python
from pathlib import Path

from Bio import SeqIO
from Bio.SeqUtils import gc_fraction
import pandas as pd
```

```python
data = Path().resolve()
```

```python
with open(data / "input.txt") as file:
    files = [line.strip() for line in file.readlines()]
```

```python
df = pd.DataFrame(columns=["accession", "size", "gc"])
for idx, file in enumerate(files):
    with open(file, "r") as handle:
        seq = SeqIO.read(handle, format="fasta")
    df.loc[idx] = [seq.id, len(seq.seq), gc_fraction(seq.seq)]
```

```python
df.to_csv(data / "contig_info.csv", index=False)
```
