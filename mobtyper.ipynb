```python
from pathlib import Path
import subprocess

import pandas as pd
```

```python
data = Path().resolve() / 'mobtyper'
```

```python
with open(data.parent / "input.txt") as file:
    files = [Path(line.strip()) for line in file.readlines()]
```

```python
for file in files:
    cmd = f"""
    source activate mob-suite \
        && bsub \
            -o {data / 'mobtyper.out'} \
            -e {data / 'mobtyper.err'} \
                "mob_typer \
                    --infile {file} \
                    --out_file {data / f'{file.stem}_mobtyper.txt'} \
                    --biomarker_report_file {data / f'{file.stem}_biomarkers.txt'} \
                "
    """
    subprocess.run(cmd, shell=True)
```

```python
df = pd.concat([pd.read_table(file, sep='\t') for file in data.glob('*mobtyper.txt')], ignore_index=True)
```

```python
df[['sample_id', 'rep_type(s)', 'relaxase_type(s)', 'mpf_type', 'orit_type(s)', 'predicted_mobility', 'predicted_host_range_overall_name']].to_csv(data / "mobtyper_merged.csv")
```
